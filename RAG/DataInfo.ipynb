```python
import pandas as pd

df = pd.read_csv("지하철 데이터.csv")

# 호선명 + 역이름 분리
df["line_name"] = df["business_name"].str.extract(r"(.*?호선)")
df["station_name"] = df["business_name"].str.replace(r".*?호선 ", "", regex=True)

# 정리된 컬럼 저장
df = df[["station_name", "line_name", "latitude", "longitude"]]
df.columns = ["station_name", "line_name", "lat", "lng"]

# 새 CSV 저장
df.to_csv("subway_stations_clean.csv", index=False)
```

```python
import pandas as pd

# 데이터 불러오기
df = pd.read_csv('subway_stations_clean.csv')

# line_name이 비어 있는 경우, station_name을 기준으로 채우기
def fix_line_name(row):
    if pd.notnull(row['line_name']):
        return row['line_name']
    
    name = row['station_name']
    if '수인분당' in name:
        return '수인분당선'
    elif '우이신설' in name:
        return '우이신설선'
    elif '경인' in name:
        return '1호선'
    elif 'GTX-A' in name:
        return 'GTX-A'
    elif 'GTX-B' in name:
        return 'GTX-B'
    elif 'GTX-C' in name:
        return 'GTX-C'
    elif '신림선' in name:
        return '신림선'
    elif '경의선' in name:
        return '경의선'
    elif '공항철도' in name:
        return '공항철도'
    elif '경춘선' in name:
        return '경춘선'
    elif '신분당선' in name:
        return '신분당선'
    else:
        return '미정'  # 최악의 경우

# 수정 적용
df['line_name'] = df.apply(fix_line_name, axis=1)

# 수정된 결과 저장
df.to_csv('subway_stations_fixed.csv', index=False)
```
