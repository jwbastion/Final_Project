```python
import pandas as pd
import requests
from tqdm import tqdm
from dotenv import load_dotenv
import os
from pathlib import Path

# 🔹 .env 파일에서 API 키 불러오기
env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)
KAKAO_API_KEY = os.getenv("KAKAO_API_KEY")

# 🔹 Kakao Reverse Geocoding 함수
def get_road_address(longitude, latitude):
    url = "https://dapi.kakao.com/v2/local/geo/coord2address.json"
    headers = {"Authorization": f"KakaoAK {KAKAO_API_KEY}"}
    params = {
        "x": longitude,  # 경도
        "y": latitude,   # 위도
    }

    response = requests.get(url, headers=headers, params=params)
    if response.status_code == 200:
        result = response.json()
        if result['documents']:
            return result['documents'][0]['road_address']['address_name'] if result['documents'][0]['road_address'] else None
    return None

# 🔹 CSV 파일 불러오기
df = pd.read_csv("C:/Users/user/Desktop/Final_Project/test/daisoo.csv")  # 또는 다른 좌표 데이터

# 🔹 도로명 주소 생성
tqdm.pandas()
df['도로명주소'] = df.progress_apply(lambda row: get_road_address(row['longitude'], row['latitude']), axis=1)

# 🔹 저장
df.to_csv("C:/Users/user/Desktop/Final_Project/test/daiso.csv", index=False, encoding="utf-8")
df[['도로명주소']].head()
```
