```python
import pandas as pd
import re

# 파일 불러오기
cinema = pd.read_csv("../data/play/cinema.csv")
pc = pd.read_csv("../data/play/pc.csv")
sing = pd.read_csv("../data/play/sing.csv")
```

```python
# 주제 추가
cinema['category'] = '영화관'
pc['category'] = 'PC방'
sing['category'] = '노래방'
```

```python
# 통합
play_df = pd.concat([cinema, pc, sing], ignore_index=True)
```

```python
# 컬럼명 변경
# land_address   : 지번 주소
# road_address   : 도로명 주소
# postal_code    : 도로명 우편번호
# business_name  : 사업장 이름
# longitude      : 경도
# latitude       : 위도
# category       : 업종 분류 (영화관 / PC방 / 노래방)
play_df.columns = ['land_address', 'road_address', 'postal_code', 'business_name', 'longitude', 'latitude', 'category']
```

```python
print(play_df.nunique()) # 고유 값 개수
```

```python
print("결측치 개수:")
print(play_df.isnull().sum())
```

위경도 부분 464개의 결측치

```python
print(play_df.sample(5)) # 샘플 데이터 확인
```

```python
# 우편번호 정제
def clean_postal_code(x):
    if pd.isna(x):
        return pd.NA
    digits = re.sub(r'[^0-9]', '', str(x))
    return digits.zfill(5) if len(digits) == 5 else pd.NA
```

```python
play_df['postal_code'] = play_df['postal_code'].apply(clean_postal_code)
```

```python
# 위도/경도 결측치 제거
play_df = play_df.dropna(subset=['longitude', 'latitude'])
```

```python
# ID 컬럼 추가
play_df.reset_index(drop=True, inplace=True)
play_df.insert(0, 'id', play_df.index + 1)
```

```python
# 정제된 결과 확인
print("\n 정제 후 데이터 개수:", len(play_df))
```

```python
# 저장
play_df.to_csv("../data/play/play.csv", index=False, encoding='utf-8-sig')
```
