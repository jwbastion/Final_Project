```python
import pandas as pd
import re

# 파일 불러오기
life = pd.read_csv("../data/life/life.csv")
```

```python
print(life)
```

```python
bigmart.drop(columns=['지번주소'], inplace=True)
department_store.drop(columns=['지번주소', 'Latitude', 'Longitude'], inplace=True)
park.drop(columns=['공원자치구'], inplace=True)
post_office.drop(columns=['H_KOR_GU', 'H_KOR_DONG'], inplace=True)
```

```python
# 주제 추가
bigmart['category'] = '대형마트'
department_store['category'] = '백화점'
park['category'] = '공원'
post_office['category'] = '우체국'
```

```python
print(bigmart.columns)
print(department_store.columns)
print(park.columns)
print(post_office.columns)
```

```python
# 예: park이라는 DataFrame에서 컬럼 순서 바꾸기
park = park[['도로명주소', '공원명', '경도', '위도', 'category']]
```

```python
park.rename(columns={'공원명': '사업자명'}, inplace=True)
```

```python
print(bigmart.columns)
print(department_store.columns)
print(park.columns)
print(post_office.columns)
```

```python
# 통합
life_df = pd.concat([bigmart, department_store, park, post_office], ignore_index=True)
```

```python
# 컬럼명 변경
# land_address   : 지번 주소
# road_address   : 도로명 주소
# postal_code    : 도로명 우편번호
# business_name  : 사업장 이름
# longitude      : 경도
# latitude       : 위도
# category       : 업종 분류 (영화관 / PC방 / 노래방)
life_df.columns = ['road_address','business_name', 'longitude', 'latitude', 'category']
```

```python
print(life_df.nunique()) # 고유 값 개수
```

```python
print("결측치 개수:")
print(life_df.isnull().sum())
```

위경도 부분 464개의 결측치

```python
print(life_df.sample(5)) # 샘플 데이터 확인
```

```python
# 위도/경도 결측치 제거
life_df = life_df.dropna(subset=['longitude', 'latitude'])
```

```python
# ID 컬럼 추가
life_df.reset_index(drop=True, inplace=True)
life_df.insert(0, 'id', life_df.index + 1)
```

```python
# 정제된 결과 확인
print("\n 정제 후 데이터 개수:", len(life_df))
```

```python
# 저장
life.to_csv("../data/life/life.csv", index=False, encoding='utf-8')
```
