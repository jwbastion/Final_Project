```python
import pandas as pd
import re

# 파일 불러오기
officetels = pd.read_csv("../data/officetels/officetels.csv")
```

```python
print(officetels)
```

```python
# 주제 추가
officetels['category'] = '오피스텔'
```

```python
# 통합
print(officetels.columns.tolist())
```

```python
print("결측치 개수:")
print(officetels.isnull().sum())
```

```python
officetels.drop(columns=['id', 'code','note', 'created_at', 'updated_at'], inplace=True)
```

```python
print(officetels.columns.tolist())
```

```python
# 컬럼명 변경
# land_address   : 지번 주소
# road_address   : 도로명 주소
# postal_code    : 도로명 우편번호
# business_name  : 사업장 이름
# longitude      : 경도
# latitude       : 위도
# category       : 업종 분류 (영화관 / PC방 / 노래방)
officetels.columns = ['건물이름', '위도', '경도', '지번주소', '도로명주소', '준공연도', '평균전세가격', '월세보증금평균', '월세평균', '지도URL', '총세대수', '최고층수', '공급면적', '임대가능최소면적', '임대가능최대면적', '근처지하철역', '최근거래층수', '최근거래전용면적', '최근거래날짜', '최근월세거래가격', '최근월세보증금거래가격', 'category']
```

```python
officetels.drop(columns=['근처지하철역', '지번주소', '최고층수', '총세대수', '지도URL', '근처지하철역', '최근거래층수', '최근거래전용면적', '최근거래날짜', '최근월세거래가격', '최근월세보증금거래가격'], inplace=True)
```

```python
print(officetels.nunique()) # 고유 값 개수
```

```python
print("결측치 개수:")
print(officetels.isnull().sum())
```

```python
# 위도/경도 결측치 제거
officetels = officetels.dropna(subset=['공급면적'])
```

```python
# ID 컬럼 추가
officetels.reset_index(drop=True, inplace=True)
officetels.insert(0, 'id', officetels.index + 1)
```

```python
# 정제된 결과 확인
print("\n 정제 후 데이터 개수:", len(officetels))
```

```python
print(officetels)
```

```python
officetels.rename(columns={'위도': 'latitude', '경도': 'longitude', '도로명주소': 'road_address'}, inplace=True)
```

```python
print(officetels.isnull().sum())
```

```python
officetels = officetels[['id', 'road_address', '건물이름', 'longitude', 'latitude', 'category']]
```

```python
# 저장
officetels.to_csv("../data/officetels/officetels123.csv", index=False, encoding='utf-8')
```
