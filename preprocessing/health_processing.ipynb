```python
import pandas as pd
import re

# 파일 불러오기
golf = pd.read_csv("../data/health/golf.csv")
health = pd.read_csv("../data/health/health.csv")
```

```python
# 주제 추가
golf['category'] = '골프장'
health['category'] = '헬스장'
```

```python
print(health)
```

```python
print(golf)
```

```python

golf.drop(columns=['지번주소', '도로명우편번호'], inplace=True)
health.drop(columns=['지번주소', '도로명우편번호'], inplace=True)
```

```python
# 통합
health_df = pd.concat([golf, health], ignore_index=True)
```

```python
print(health_df)
```

```python
# 컬럼명 변경
# land_address   : 지번 주소
# road_address   : 도로명 주소
# postal_code    : 도로명 우편번호
# business_name  : 사업장 이름
# longitude      : 경도
# latitude       : 위도
# category       : 업종 분류 (영화관 / PC방 / 노래방)
health_df.columns = ['road_address', 'business_name', 'longitude', 'latitude', 'category']
```

```python
print(health_df.nunique()) # 고유 값 개수
```

```python
print("결측치 개수:")
print(health_df.isnull().sum())
```

위경도 부분 464개의 결측치

```python
print(health_df)
```

```python
print(health_df.sample(5)) # 샘플 데이터 확인
```

```python
# 위도/경도 결측치 제거
health_df = health_df.dropna(subset=['longitude', 'latitude'])
```

```python
# ID 컬럼 추가
health_df.reset_index(drop=True, inplace=True)
health_df.insert(0, 'id', health_df.index + 1)
```

```python
# 정제된 결과 확인
print("\n 정제 후 데이터 개수:", len(health_df))
```

```python
# 저장
health_df.to_csv("../data/health/health.csv", index=False, encoding='utf-8')
```
